# ANN Lab 2
## Single Neuron for Student Placement Prediction

```python
import numpy as np
```

```python
# Sigmoid Activation Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
# Training Data
# Features:
# CGPA
# Internship (1=Yes, 0=No)
# Communication Skill (1=Good, 0=Poor)

X = np.array([
    [6.5,0,0],
    [7.8,1,1],
    [8.5,1,1],
    [5.9,0,0],
    [7.2,1,0],
    [6.8,0,1],
    [8.9,1,1],
    [5.8,0,0]
])

y = np.array([0,1,1,0,1,0,1,0])
```

```python
# Normalize CGPA
X[:,0] = X[:,0] / 10
```

```python
# Initialize Parameters
weights = np.zeros(3)
bias = 0
learning_rate = 0.5
epochs = 1000
```

```python
# Training
for epoch in range(epochs):
    for i in range(len(X)):
        x = X[i]
        target = y[i]

        net = np.dot(x, weights) + bias
        output = sigmoid(net)

        error = target - output

        weights += learning_rate * error * x
        bias += learning_rate * error
```

```python
# Testing
print('Final Weights:', weights)
print('Final Bias:', bias)

print('\nPredictions:')
for x in X:
    probability = sigmoid(np.dot(x, weights) + bias)
    prediction = 1 if probability >= 0.5 else 0
    print(x, '->', prediction, '| Probability =', round(probability,3))
```
